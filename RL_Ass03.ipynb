# Modified ε-Greedy Strategy for Multi-Armed Bandit

## Objective
The objective is to implement the modified ε-greedy strategy for a Multi-Armed Bandit problem and estimate the future reward (Q-value) of each machine using the incremental sample-average method.

## Given
*   **Number of machines**: `M = 3`
*   **Machine names**: `machines = ['A', 'B', 'C']`
*   **Number of plays**: `N_play = 10`
*   **Exploration probability**: `ϵ = 0.2`
*   **Initial values**:
    *   `N = [0, 0, 0]`
    *   `R = [0, 0, 0]`
    *   `Q = [0, 0, 0]`

## ε-Greedy Strategy
For every play, generate a random number `r` between 0 and 1.

*   **If**: `r < ϵ`
    Then **exploration** is performed. In exploration, randomly select one machine *except* the machine having the highest Q-value.
*   **Otherwise**:
    **Exploitation** is performed. The machine having the highest Q-value is selected.

## Incremental Sample-Average Formula
The Q-value is updated using:

`Q_new(a) = Q_old(a) + (R - Q_old(a)) / N(a)`

Where:
*   `Q(a)` = current estimated value of machine `a`
*   `R` = reward obtained
*   `N(a)` = number of times machine `a` has been selected

```python
import random

# Part A - Machine Initialization
machines = ['A', 'B', 'C']

M = 3
N_play = 10
epsilon = 0.2

# Part B - Initialize tables
N = [0, 0, 0]
R = [0, 0, 0]
Q = [0.0, 0.0, 0.0]

total_reward = 0

for play in range(1, N_play + 1):

    # Part C - Modified Epsilon-Greedy Decision
    r = random.random()

    # Find machine with highest Q-value
    highest_Q = max(Q)
    best_machine = Q.index(highest_Q)

    if r < epsilon:
        # Exploration:
        # randomly select a machine except best machine
        choices = [i for i in range(M) if i != best_machine]
        selected = random.choice(choices)
        decision = "Exploration"
    else:
        # Exploitation:
        # select machine with highest Q-value
        selected = best_machine
        decision = "Exploitation"

    print("\nPlay", play)
    print("Random number =", round(r, 4))
    print("Decision =", decision)
    print("Play with machine", machines[selected])

    # Part D - Reward Collection
    reward = float(input("Enter reward: "))

    # Part E - Update N, R and Q
    N[selected] += 1
    R[selected] += reward

    Q[selected] = Q[selected] + (
        reward - Q[selected]
    ) / N[selected]

    total_reward += reward

# Part F - Display Results
print("\n================================")
print("        FINAL RESULTS")
print("================================")

print("Machine\tN\tR\tQ-value")

for i in range(M):
    print(
        machines[i], "\t",
        N[i], "\t",
        R[i], "\t",
        round(Q[i], 4)
    )

print("\nTotal Reward =", total_reward)
```
