```python
import math
machines = ['A', 'B', 'C']

M = 3
N_play = 10
```

```python
#Number of times each machine is played
N = {machine: 0 for machine in machines}

# Rewards received from each machine
R = {machine: [] for machine in machines}

# Estimated Q-values
Q = {machine: 0.0 for machine in machines}

# UCB values
UCB = {machine: float('inf') for machine in machines}

# Total reward
total_reward = 0
```

```python
def calculate_ucb(machine, t):

    # If machine has not been played,
    # give infinite UCB value for exploration
    if N[machine] == 0:
        return float('inf')

    ucb_value = Q[machine] + math.sqrt(
        (2 * math.log(t)) / N[machine]
    )

    return ucb_value
```

```python
def calculate_ucb(machine, t):

    # If machine has not been played,
    # give infinite UCB value for exploration
    if N[machine] == 0:
        return float('inf')

    ucb_value = Q[machine] + math.sqrt(
        (2 * math.log(t)) / N[machine]
    )

    return ucb_value

for play in range(1, N_play + 1):

    print("\n" + "=" * 40)
    print("Play Number:", play)

    # Calculate UCB for every machine
    for machine in machines:
        UCB[machine] = calculate_ucb(machine, play)

    # Display UCB values
    print("\nUCB Values:")

    for machine in machines:
        if UCB[machine] == float('inf'):
            print(machine, ": Infinity")
        else:
            print(machine, ":", round(UCB[machine], 3))

    # Select machine with highest UCB
    selected_machine = max(UCB, key=UCB.get)

    print("\nPlay with Machine:", selected_machine)

    # Get reward from user
    reward = float(input("Enter reward: "))

    # Store reward
    R[selected_machine].append(reward)

    # Update number of times selected
    N[selected_machine] += 1

    # Incremental Sample Average Q-value Update
    Q[selected_machine] = Q[selected_machine] + (
        (reward - Q[selected_machine]) / N[selected_machine]
    )

    # Update total reward
    total_reward += reward

    print("\nUpdated Q Values:")
    for machine in machines:
        print(machine, ":", round(Q[machine], 3))
```
